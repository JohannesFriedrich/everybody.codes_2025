# Part 1

```python
file_name = "everybody_codes_e2025_q09_p1.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

DNA = []
for d in data:
    x = d.split(":")
    DNA.append(x[1])

# find child index
def find_child(l1, l2, l3):
    pos = 0
    is_child = [True] * 3
    child_index = -1
    while True:
        chars = [l1[pos], l2[pos], l3[pos]]
        unique = [c for c in chars if chars.count(c) == 1]
        if len(unique) != 0:
            index = chars.index(unique[0])
            is_child[index] = False
        if is_child.count(True) == 1:
            child_index = [i for i ,v in enumerate(is_child) if v is True][0]
            break
        pos += 1
    return child_index

child_index = find_child(DNA[0], DNA[1], DNA[2])

def calc_DOS(child_index, parent_index1, parent_index2):
    p1 = 0
    p2 = 0
    for pos in range(len(DNA[child_index])):
        p1 += DNA[child_index][pos] == DNA[parent_index1][pos]
        p2 += DNA[child_index][pos] == DNA[parent_index2][pos]

    return p1*p2

calc_DOS(child_index, (child_index+1)%3, (child_index-1)%3)
```

# Part 2

```python
import itertools

file_name = "everybody_codes_e2025_q09_p2.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

DNA = {}
for d in data:
    x = d.split(":")
    DNA[x[0]] = x[1]

def find_child(l1, l2, l3):
    pos = 0
    is_child = [True] * 3
    child_index = -1
    while pos < len(l1):
        chars = [l1[pos], l2[pos], l3[pos]]
        unique = [c for c in chars if chars.count(c) == 1]
        if len(unique) != 0:
            index = chars.index(unique[0])
            is_child[index] = False
        if is_child.count(True) == 0:
            return -1
        pos += 1
        
    child_index = [i for i ,v in enumerate(is_child) if v is True][0]
    return child_index

combos = list(itertools.combinations(DNA, 3))
res = 0
for combo in combos:
    i1, i2, i3 = combo
    # print(i1,i2,i3)
    child_index = find_child(DNA[i1], DNA[i2], DNA[i3])
    if child_index != -1:
        # print("Found:", i1, i2, i3)
        res += calc_DOS(combo[child_index], combo[(child_index+1)%3], combo[(child_index-1)%3])

print(res)
```
