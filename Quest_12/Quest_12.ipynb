```python
from collections import deque
import copy

file_name = "everybody_codes_e2025_q12_p3.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")
  
G = [[int(c) for c in line] for line in data]
R = len(G)
C = len(G[0])

def find_highest_number(G, max=10) -> int:

	highest_number = -1
	for r in range(len(G)):
		for c in range(len(G[0])):
			if G[r][c] > highest_number and G[r][c] < max:
				highest_number = G[r][c]

	return highest_number


def bfs(start_field: tuple, G) -> set:

	queue = deque()
	queue.append(start_field)
	exploded = set()
	while queue:

		(r,c) = queue.popleft()

		if (r,c) in exploded:
			continue
		exploded.add((r,c))

		# find neighbours
		for dr,dc in [(-1,0),(0,1),(1,0),(0,-1)]:
			rr = r+dr
			cc = c+dc
			if 0<=rr<R and 0<=cc<C and G[rr][cc] <= G[r][c] and (rr,cc) not in exploded and G[r][c] != "#":
				queue.append((rr,cc))	

	return exploded

def simulate_round(G):

	max_explosion = 0
	for r in range(R):
		for c in range(C):
			if G[r][c]<10:
				exploded = bfs((r,c), G)
				if len(exploded) > max_explosion:
					max_explosion = len(exploded)
					G_final = copy.deepcopy(G)
					for x,y in exploded:
						G_final[x][y] = 10

	print(max_explosion)
	return G_final

round1 = simulate_round(G)
round2 = simulate_round(round1)
round3 = simulate_round(round2)

res = 0
for r in range(R):
		for c in range(C):
			if round3[r][c] == 10:
				res += 1
res
```
