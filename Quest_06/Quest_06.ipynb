# Part 1

```python
file_name = "everybody_codes_e2025_q06_p1.txt"
file = open(file_name, "r")
data = file.read().strip()

def get_mentors(data, letter):
    
    lower_positions = [i for i, c in enumerate(data) if c == letter.lower()]
    upper_positions = [i for i, c in enumerate(data) if c == letter.upper()]

    res = 0
    for lower in lower_positions:
        for upper in upper_positions:
            if upper < lower:
                res += 1
    return res

get_mentors(data, "a")
```
