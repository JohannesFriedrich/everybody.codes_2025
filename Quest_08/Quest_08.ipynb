```python
import math
from collections import defaultdict

file_name = "everybody_codes_e2025_q08_p2.txt"
# file_name = "test.txt"
file = open(file_name, "r")
data = [int(x) for x in file.read().strip().split(",")]

def line_intersect(a: int, b: int, c: int, d:int ) -> bool:

    a, b = sorted((a, b))
    c, d = sorted((c, d))

    # print(a,b,c,d)
    # if two points are identical: no intersection inside circle
    if b == c or b == d or a == c or a == d:
        return False

    # Endpunkte von (c,d) liegen genau dann „abwechselnd“ bzgl. (a,b),
    # wenn genau einer von c,d im offenen Intervall (a,b) liegt:
    c_in = a < c < b
    d_in = a < d < b

    return c_in ^ d_in

# print(line_intersect(3,5,3,4))

ropes = [(data[i], data[i+1]) if data[i] < data[i+1] else (data[i+1], data[i]) for i in range(0, len(data) - 1)]
distinct = list(set(ropes))

print(distinct)

lookup = defaultdict(list)

for p1 in distinct:
    for p2 in distinct:
        if p1 != p2:
            lookup[f"{p1[0]}{p1[1]}"].append({f"{p2[0]}{p2[1]}": line_intersect(p1[0], p1[1], p2[0], p2[1])})


# for p1 in range(1,257):
#     print(f"Processing p1={p1}")
#     for p2 in range(1,257):
#         # print(f"Processing p2={p2}")
#         if p2 > p1:
#             for p3 in range(1,257):
#                 for p4 in range(1,257):
#                     if p4 > p3:
#                         lookup[f"{p1}{p2}"].append({f"{p3}{p4}": line_intersect(p1, p2, p3, p4)})



# res = 0
# for i, rope in enumerate(ropes):
#     for other_rope in ropes[i+1:]:
#         if line_intersect(rope[0], rope[1], other_rope[0], other_rope[1]):
#             res += 1

# res
```
