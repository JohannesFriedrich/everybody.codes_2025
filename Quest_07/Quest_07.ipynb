# Part 1

```python
file_name = "everybody_codes_e2025_q07_p1.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

names = data[0].split(",")
rules = {}
for line in data[2:]:
    l1, l2 = line.split(" > ")
    rules[l1] = l2.split(",")

def check_rules(names, rules):

    names_res = {}
    for name in names:
        names_res[name] = True
        found = 0
        for i in range(len(name) - 1):
            letter_combo = name[i:i+2]
            for letter, rule in rules.items():
                for r in rule:
                    if letter + r not in letter_combo:
                        continue
                    else:
                        found += 1
        if found != len(name) - 1:
            names_res[name] = False

    return names_res
       

[name for name, res in check_rules(names, rules).items() if res is True]

```

```python
file_name = "everybody_codes_e2025_q07_p2.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

names = data[0].split(",")
rules = {}
for line in data[2:]:
    l1, l2 = line.split(" > ")
    rules[l1] = l2.split(",")

i = 1
res = 0
for v in check_rules(names, rules).values():
    if v:
        res += i
    i += 1
print(res)

```
