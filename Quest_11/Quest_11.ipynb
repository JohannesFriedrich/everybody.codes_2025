# Part 1

```python
file_name = "everybody_codes_e2025_q11_p1.txt"
# file_name = "test.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

data = [int(i) for i in data]

# 1) if next col has less -> move to next col
def get_next_round(data, phase = 1):
    for i in range(len(data)-1):
        if phase == 1:
            if data[i+1] < data[i]:
                data[i] -= 1
                data[i+1] += 1
        if phase == 2:
            if data[i+1] > data[i]:
                data[i] += 1
                data[i+1] -= 1
    return data

round = 0
while True:
    round += 1
    last_round = data.copy()
    data = get_next_round(data,phase=1)
    print(data)
    if last_round == data:
        break
# set one round back
round -= 1

while round<10:
    round += 1
    data = get_next_round(data,phase=2)
    print(data)

# get flock checksum:
sum([(i+1)*v for i,v in enumerate(data)])


```

# Part 3

```python
file_name = "everybody_codes_e2025_q11_p3.txt"
# file_name = "test.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

data = [int(i) for i in data]

# 1) if next col has less -> move to next col
def get_next_round(data, phase = 1):
    for i in range(len(data)-1):
        if phase == 1:
            if data[i+1] < data[i]:
                data[i] -= 1
                data[i+1] += 1
        if phase == 2:
            if data[i+1] > data[i]:
                data[i] += 1
                data[i+1] -= 1
    return data

round = 0
while True:
    round += 1
    last_round = data.copy()
    data = get_next_round(data,phase=1)
    if last_round == data:
        break
# set one round back
round -= 1

# calculate mean because all cols end here. for all cols smaller then the mean we need the difference
# between this col and the mean. if a col is bigger it gets reduced automatically.
mean = int(sum(data)/len(data))

sum([mean-i for i in data if i<mean])

```
