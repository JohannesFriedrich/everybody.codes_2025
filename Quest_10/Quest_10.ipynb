# Part 1

```python
from collections import deque

file_name = "everybody_codes_e2025_q10_p1.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

G = [[c for c in line] for line in data]
R = len(G)
C = len(G[0])

sheep_field = []
for r in range(R):
    for c in range(C):
        if G[r][c] == "D":
            dragon_start = (r,c,0)
        if G[r][c] == "S":
            sheep_field.append((r,c))

visited = set()
queue = deque()
queue.append(dragon_start)
res = 0 
moves = 4

while queue:
    r, c, m  = queue.popleft()

    if m > moves-1:
        break 

    if (r,c) in visited:
        continue
    visited.add((r,c))

    for dr, dc in [(2,1), (2,-1), (-2,1), (-2,-1), (1,2), (-1,2), (1,-2), (-1,-2)]:
        rr = r + dr
        cc = c + dc
        if 0 <= rr < R and 0 <= cc < C and (rr,cc) not in visited:
            # print(f"Dragon jumped to {(rr,cc)}")
            if (rr,cc) in sheep_field:
                sheep_field.remove((rr,cc))
                # print(f"Sheep in {(rr,cc)} found")
                res += 1
            queue.append((rr,cc,m+1))

print(res)
```

```python
from collections import deque

file_name = "everybody_codes_e2025_q10_p2.txt"
file = open(file_name, "r")
data = file.read().strip().split("\n")

G = [[c for c in line] for line in data]
R = len(G)
C = len(G[0])

sheep_field_orig = []
hideouts = []
for r in range(R):
    for c in range(C):
        if G[r][c] == "D":
            dragon_start = (r,c,0)
        if G[r][c] == "S":
            sheep_field_orig.append((r,c))
        if G[r][c] == "#":
            hideouts.append((r,c))

min_row_sheeps = min([v[0] for v in sheep_field_orig])

def get_new_sheep_positions(sheep_field, shift):
    temp = []
    for v in sheep_field:
        temp.append((v[0]+shift, v[1]))
    return temp

```

```python
visited = set()
queue = deque()
queue.append(dragon_start)
res = 0 
moves = 20
m_max = -1

while queue:
    r, c, m  = queue.popleft()

    if m>m_max:
        m_max = m
        print(f"At move {m_max} with {res} sheep caught. Sheep left: {len(sheep_field_orig)}")
        print(f"Len queue: {len(queue)}")
    if m > moves-1:
        break 


    for dr, dc in [(2,1), (2,-1), (-2,1), (-2,-1), (1,2), (-1,2), (1,-2), (-1,-2)]:
        rr = r + dr
        cc = c + dc
        if 0 <= rr < R and 0 <= cc < C:
            # Dragons move
            sheep_field = get_new_sheep_positions(sheep_field_orig, m)
            if (rr,cc) in sheep_field and G[rr][cc] != "#":
                sheep_field_orig.remove((rr-m,cc))
                res += 1

            # Sheeps move
            sheep_field = get_new_sheep_positions(sheep_field_orig, m+1)
            # print(f"Sheeps move: {sheep_field}")
            if (rr,cc) in sheep_field and G[rr][cc] != "#":
                sheep_field_orig.remove((rr-m-1,cc))
                res += 1
            
            if (rr,cc, m+1) not in queue:
                queue.append((rr, cc, m+1))



print(res)
```
